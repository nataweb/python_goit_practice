#### Який відсоток респондентів програмує на Python?

```python
total =len(df)
python_count = df["LanguageHaveWorkedWith"].str.contains("Python", na=False).sum()
python_percent = python_count / total * 100

print("Кількість Python-програмістів:", python_count)
print("Відсоток респондентів, що програмують на Python:", python_percent)
```

#### Скільки респондентів навчалося програмувати за допомогою онлайн курсів?

#### Серед респондентів що програмують на Python в групуванні по країнам, яка середня та медіанна сума компенсації (ConvertedCompYearly) в кожній країні?

```python
python_df = df[df["LanguageHaveWorkedWith"].str.contains("Python", na=False)]
comp_stats = python_df.groupby("Country")["ConvertedCompYearly"].agg(["mean", "median"])
print(comp_stats)
```
